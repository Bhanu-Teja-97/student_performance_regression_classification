Handling the all Numerical data by using median()


```python
df['math_score'] = df['math_score'].fillna(df['math_score'].median())
df['reading_score'] = df['reading_score'].fillna( df['reading_score'].median())
df['writing_score'] = df['writing_score'].fillna(df['writing_score'].median())
df['attendance_pct'] = df['attendance_pct'].fillna(df['attendance_pct'].median())
df['study_hours_per_week'] = df['study_hours_per_week'].fillna(df['study_hours_per_week'].median())
df.info()
```

```python
df.info()
```

NOW using mode() for the cat


```python
df['parental_education'] = df['parental_education'].fillna( df['parental_education'].mode()[0])
df['extracurricular'] = df['extracurricular'].fillna(df['extracurricular'].mode()[0])
df['pass_final'] = df['pass_final'].fillna(df['pass_final'].mode()[0])
df.info()

```

```python
df.isnull().sum()
```

NOW we are removing duplicate col to get the analysis more accurate
. now we can see that col are reduce from 12 to 9





```python
df = df.drop(columns=[
    'math_score.1',
    'math_score.2',
    'math_score.3'
])
```

```python
df.info()
```

checking the unique values in col using the keyword unique()

```python
df["parental_education"].unique()
```

```python
df['parental_education'] = df['parental_education'].replace({

    # Bachelors
    'BACHELORS': 'Bachelors',
    'Bachlors': 'Bachelors',
    'Bachelor': 'Bachelors',
    'Bachelos': 'Bachelors',
    'Bachelrs': 'Bachelors',
    'Bahelors': 'Bachelors',
    'Bacelors': 'Bachelors',
    'achelors': 'Bachelors',

    # Masters
    'MASTERS': 'Masters',
    'Master': 'Masters',
    'Masers': 'Masters',
    'Msters': 'Masters',
    'Maters': 'Masters',
    'Mastrs': 'Masters',

    # PhD
    'Phd': 'PhD',
    'hD': 'PhD',
    'PHD': 'PhD',
    'PD': 'PhD',
    'Ph': 'PhD',

    # High School
    'HIGH SCHOOL': 'High School',
    'High school': 'High School',
    'High chool': 'High School',
    'Hgh School': 'High School',
    'Hih School': 'High School'
})
```

```python
print("Missing IDs:", df['student_id'].isnull().sum())
print(df['student_id'].dropna().head())
```

```python
missing = df['student_id'].isnull()

existing_nums = df['student_id'].dropna().str.extract(r'(\d+)')[0].astype(int)

next_id = existing_nums.max() + 1

df.loc[missing, 'student_id'] = [
    f'S{i}' for i in range(next_id, next_id + missing.sum())
]
```

3. Data analysis



1. Univariate Analysis

   Graph 1: Math score distribution (Histogram)

   From the graph most of the students nearly 60 students score 70+marks in math thats the highest mark percentage

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df1['math_score'], bins=15, kde=True)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

Graph 2: Reading score distribution (Boxplot)
in this box plot we observe that the reading score is staring from 60 to 75
most of the mean score is near 70

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df1['reading_score'])
plt.title('Distribution of Reading Scores')
plt.show()
```

Graph 3: Parental education distribution (Countplot)

```python
plt.figure(figsize=(8,5))
sns.countplot(x='parental_education', data=df1,color='orange')
plt.title('Parental Education Distribution')
plt.xticks(rotation=30)
plt.show()
```

Graph 4: Extracurricular participation (Pie chart)

```python
df['extracurricular'].value_counts().plot( kind='pie',autopct='%1.1f%%',figsize=(6,6))
plt.title('Extracurricular Participation')
plt.ylabel('')
plt.show()
```

2. Bivariate Analysis

Graph 1: Math score vs Reading score (Scatter plot)


```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df1, x='math_score', y='reading_score')
plt.title('Math Score vs Reading Score')
plt.show()
```

Graph 2: Study hours vs Math score (Regression plot)

```python
plt.figure(figsize=(8,5))
sns.regplot(data=df1, x='study_hours_per_week', y='math_score')
plt.title('Study Hours vs Math Score')
plt.show()
```

Graph 3: Parental education vs Writing score (Boxplot)

```python
df1.groupby('parental_education')['writing_score'].describe()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df1, x='parental_education', y='writing_score')
plt.title('Writing Score After Outlier Removal')
plt.xticks(rotation=30)
plt.show()
```

Graph 4: Extracurricular vs Pass/Fail (Countplot)

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df1, x='extracurricular', hue='pass_final')
plt.title('Extracurricular vs Pass/Fail')
plt.show()
```

3. Multivariate Analysis
Graph 1: Correlation heatmap

```python
plt.figure(figsize=(9,6))
sns.heatmap(
    df1.select_dtypes(include='number').corr(),annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Student Performance Correlation Heatmap')
plt.show()
```

Graph 2: Pairplot

```python
sns.pairplot(
    df1,vars=['math_score', 'reading_score','writing_score', 'study_hours_per_week'], hue='pass_final', diag_kind='hist')

plt.show()
```

Graph 3: Student scores by pass/fail (Violin plot)

```python
plt.figure(figsize=(10,6))

score_data = df1.melt( id_vars='pass_final', value_vars=['math_score', 'reading_score', 'writing_score'],var_name='Subject',value_name='Score')

sns.violinplot(
    data=score_data,
    x='Subject',
    y='Score',
    hue='pass_final',
    split=True
)

plt.title('Student Score Distribution by Pass/Fail')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
```
